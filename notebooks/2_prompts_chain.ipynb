### Prompt Templates - Dynamic Prompts

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")
```

```python
prompts = [
    {"role": "system", "content": "You are a translator and translate input in {language}."},
    {"role": "user", "content": "{query}"}
]
```

```python
prompt_template = ChatPromptTemplate.from_messages(prompts)

translation_chain = prompt_template | llm

res = translation_chain.invoke({
    "language": "Spanish",
    "query": "Hello, how are you doing today?"
})

print(res.content)
```
