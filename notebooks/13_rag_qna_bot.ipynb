```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_google_genai import GoogleGenerativeAIEmbeddings, ChatGoogleGenerativeAI
from langchain_chroma import Chroma
from langchain_core.prompts import PromptTemplate
```

```python
# Step 1 - Document Loading
loader = PyPDFLoader("../data/data_science_syllabus.pdf")
docs = loader.load()
len(docs)
```

```python
# Step 2 - Text Splitting - Chunking
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)
splitted_docs = splitter.split_documents(docs)
len(splitted_docs)
```

```python
# Step 3 - Create Vector Embeddings
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview")
```

```python
# Step 3 - Create VectorDB
vector_store = Chroma.from_documents(
    documents=splitted_docs,
    embedding=embeddings
)
```

```python
query = "Machine Learning & Data Science"
data = vector_store.similarity_search(query=query)
```

```python
context = ""
for doc in data:
    context += doc.page_content + "\n"
    
print(context)
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")
```

### Chain - Context Generator | Prompt | LLM | strparser

```python
def get_context(query:str):
    data = vector_store.similarity_search(query=query)
    context = ""
    for doc in data:
        context += doc.page_content + "\n"
    return {
        "context": context,
        "question": query
            }
```

```python
prompt = PromptTemplate.from_template("""
                        You are a helpful assistant that provides answers based on the context provided. 
                        And if you don't know the answer, just say "I don't know" and don't try to make up an answer. 
                        Context: {context}
                        Question: {question}
""")                        
```

```python
rag_chain = get_context | prompt | llm
```

```python
res = rag_chain.invoke("What is the duration of course?")
```

```python
print(res.text)
```
