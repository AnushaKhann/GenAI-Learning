```python
from pydantic import BaseModel, Field

class ResponseStructured(BaseModel):
    name: str = Field(description="The name of the person")
    profession: str = Field(description="The profession of the person")
    learning: str = Field(description="What the person is currently learning")

structured_llm = llm.with_structured_output(ResponseStructured)
```

```python
res = structured_llm.invoke(f"Give only my name and profession and what am I learning: {text}")
```
