# Text Splitter - Phase 2

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

texts = "Today, I want to speak from the heart about three pillars that define the future of our nation: " \
"our youth, our education system, and our farmers."
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=20)
```

```python
chunks = splitter.split_text(texts)
```

```python
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader("../data/speech.txt")
texts = loader.load()

# This is for a text file you can load pdf as well using PyPDFLoader
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
    )
```

```python
text_chunks = splitter.split_documents(texts)
len(text_chunks)
```
