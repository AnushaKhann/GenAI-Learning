```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = [
    "Machine Learning is a field of computer science that gives computers the ability to learn without being explicitly programmed.",
    "Artificial Intelligence is the simulation of human intelligence processes by machines, especially computer systems.",
    "Deep Learning is a subset of machine learning that uses neural networks with many layers to analyze various factors of data.",
    "Python is a high-level, interpreted programming language known for its readability and versatility.",
    "AI models can be used for a variety of applications, including natural language processing, computer vision, and predictive analytics.",
    "India won the 1983 Cricket World Cup, defeating the West Indies in the final match.",
]
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview")
```

```python
vector = embeddings.embed_documents(documents)
```

```python
len(vector)
```

```python
len(vector[0])
```

### Store Vector Embeddings in VectorDB

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(
    texts=documents,
    embedding=embeddings,
    persist_directory="./vector_db"
)
```

```python
query = "Human Intelligence"
result = vector_store.similarity_search(query=query, k=2)
```
