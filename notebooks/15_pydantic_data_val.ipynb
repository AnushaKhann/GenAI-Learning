```python
class User:
    name = ""
    email = ""
    age = 0

user = User()
user.name = "John Doe"
user.email = "john@gmail.com"
user.age = "30" # Age should be an integer. Python does not throws error here. Hence, we use Pydantic.

print(user)
```

```python
user.age # String is being assigned to an integer field.
```

```python
from pydantic import BaseModel
```

```python
class User(BaseModel):
    name: str
    email: str
    age: int
```

```python
user = User(name="John Doe", email="john@gmail.com", age=30)
user
```

### Field Validation

```python
from pydantic import Field
```

```python
class Product(BaseModel):
    title:str = Field(max_length=10)
    description:str = Field(max_length=30)
    price:float = Field(gt=0, le=100) # gt means greater than 0, le means less that or equal to 100
```

```python
product = Product(title="Laptop", description="A high-end gaming laptop", price=50)
```

```python
class Order(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    product: Product
    quantity: int = Field(gt=0, le=10)
```

```python
order = Order(id="123456", product=product, quantity=2)
order
```

```python
order.product.description
```

```python
product2 = Product(title="Smartphone", description="A high-end gaming smartphone", price=80)
```

```python
from typing import List
```

```python
class Carts(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    items: List[Product]
    comments: str = Field(max_length=100, default="No comments")

cart = Carts(id="dbjdjn", items=[product, product2])
```

```python
print(cart)
print((cart.items[0].title))
print(cart.comments)
```

```python
from typing import Annotated
```

```python
commonValidator = Annotated[List[Product], Field(default=[])]
```

```python
class Carts(BaseModel):
    id:str = Field(min_length=6, max_length=12)
    items: commonValidator
    comments: str = Field(max_length=100, default="No comments")
```

```python
class Order(BaseModel):
    product: commonValidator
```

```python
ord = Order()
ord
```

```python
ord = Order(product=[product, product2])
ord
```
