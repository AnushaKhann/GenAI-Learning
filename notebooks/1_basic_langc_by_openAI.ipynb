### Load Env Variables

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
```

```python
res = llm.invoke("Who is PM of India?")
res.content
```

### Static prompts

```python
from langchain_core.output_parsers import StrOutputParser
chain = llm | StrOutputParser() # to extract only the text string

prompts = [
    ("system", "You are a python developer."),  ### (type, content)    type -> "System", "AI", "User"
    ("user", "How to sort the array.")
]
res = chain.invoke(prompts)
```
