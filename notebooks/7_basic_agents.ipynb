```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a: int, b: int) -> int:
    """
        Add two numbers together.
        Args:
            a (int): The first number.
            b (int): The second number.
    """
    return a + b

@tool
def multiply_numbers(a: int, b: int) -> int:
    """
        Multiply two numbers together.
        Args:
            a (int): The first number.
            b (int): The second number.
    """
    return a * b
```

```python
add_numbers.invoke({"a": 5, "b": 10})
```

```python
multiply_numbers.invoke({"a": 5, "b": 10})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers, multiply_numbers],
    system_prompt="You are a math teacher and always use tools for calculations."
)
```

```python
response = agent.invoke({"messages": [{"role": "user", "content": "What is the sum of 5 and 10?"}]})
```

```python
print(response["messages"][-1].text)
```

```python
response = agent.invoke({"messages": [{"role": "user", "content": "What is the sum and product of 5 and 10?"}]})
```

```python
print(response["messages"][-1].text)
```
